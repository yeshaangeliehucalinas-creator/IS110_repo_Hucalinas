## EXERCISES

#### Exercise 1

Rewrite your pay computation with time-and-a-half for overtime and create a function called compute pay which takes two parameters (hours and rate).

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0
    
[PEW] E4.6

```python
def computepay(hours, rate):
    """Calculates pay with 1.5x rate for hours over 40."""
    if hours > 40:
        reg_pay = 40 * rate
        overtime_hours = hours - 40
        overtime_pay = overtime_hours * (rate * 1.5)
        total_pay = reg_pay + overtime_pay
    else:
        total_pay = hours * rate

    return total_pay

# --- Main Program Execution ---
try:
    # Prompt user for input
    raw_hours = input("Enter Hours: ")
    raw_rate = input("Enter Rate: ")

    # Convert inputs to floats
    f_hours = float(raw_hours)
    f_rate = float(raw_rate)

    # Call the function and store the result
    pay = computepay(f_hours, f_rate)

    # Print the result
    print(f"Pay: {pay}")

except ValueError:
    print("Error: Please enter numeric input")
```

## CHALLENGE

#### Challenge

Write a function called computegrade that takes a score as its parameter and returns a grade as a string using the following table:

    Score           Grade
    [ 98, 100]      1.0
    [ 94,  97)		1.25
    [ 90,  93)		1.5
    [ 86,  89)		1.75
    [ 82,  85)		2.0
    [ 78,  81)		2.25
    [ 75,  77)		2.5
    [ 0,   74)		5.0

    Run the program repeatedly to test the various different values for input.
    
Adapted from [PEW] E4.6
<br>

```python
def computegrade(score):
    """Returns the grade as a string based on the input score."""
    if score > 100 or score < 0:
        return "Bad score"
    elif score >= 98:
        return "1.0"
    elif score >= 94:
        return "1.25"
    elif score >= 90:
        return "1.5"
    elif score >= 86:
        return "1.75"
    elif score >= 82:
        return "2.0"
    elif score >= 78:
        return "2.25"
    elif score >= 75:
        return "2.5"
    else:
        return "5.0"

# --- Main Test Loop ---
print("Enter scores to test. Type 'exit' or 'q' to stop.")

while True:
    user_input = input("Enter score: ")

    # Allow the user to quit the loop
    if user_input.lower() in ['exit', 'q', 'quit']:
        print("Goodbye!")
        break

    try:
        # Convert input to float and call the function
        f_score = float(user_input)
        grade = computegrade(f_score)
        print(f"Grade: {grade}")

    except ValueError:
        print("Bad score (Please enter a number)")
```
