#### Exercise 2

Write your salary computation to give an employee 1.5 times the hourly rate for hours worked above 40 hours. Example:

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0
    
[PEW] E3.2

````python
# Get input from the user
try:
    hours = float(input("Enter Hours: "))
    rate = float(input("Enter Rate: "))

    if hours > 40:
        # Calculate regular pay for the first 40 hours
        reg_pay = 40 * rate
        # Calculate overtime pay for the extra hours at 1.5x rate
        overtime_hours = hours - 40
        overtime_pay = overtime_hours * (rate * 1.5)

        total_pay = reg_pay + overtime_pay
    else:
        # Standard calculation if no overtime
        total_pay = hours * rate

    print(f"Pay: {total_pay}")

except ValueError:
    print("Error: Please enter a numeric value for hours and rate.")
````

## CHALLENGE

Write a program to prompt for a score between 0 and 100 and print a grade using the following table:

    Score           Grade
    [ 98, 100]      1.0
    [ 94,  97)		1.25
    [ 90,  93)		1.5
    [ 86,  89)		1.75
    [ 82,  85)		2.0
    [ 78,  81)		2.25
    [ 75,  77)		2.5
    [ 0,   74)		5.0

Adapted from [PEW] E3.3
<br>

````python
try:
    # Prompt the user and convert to float
    input_score = input("Enter score (0-100): ")
    score = float(input_score)

    # Validate the range
    if score < 0 or score > 100:
        print("Error: Score must be between 0 and 100.")
    # Assign grades based on the table
    elif score >= 98:
        print("Grade: 1.0")
    elif score >= 94:
        print("Grade: 1.25")
    elif score >= 90:
        print("Grade: 1.5")
    elif score >= 86:
        print("Grade: 1.75")
    elif score >= 82:
        print("Grade: 2.0")
    elif score >= 78:
        print("Grade: 2.25")
    elif score >= 75:
        print("Grade: 2.5")
    else:
        print("Grade: 5.0")

except ValueError:
    print("Error: Please enter a valid numerical score.")
````
