Investigate on the difference between `is` and `==` given the following variables.
```python
a = 2
b = 2.0
c = a
```

== for almost all comparisons (numbers, strings, lists). Only use is when you are checking against singletons like None (e.g., if x is None:) or True/False.

````python
a = 2
b = 2.0
c = a

# 1. Comparing a and b
print(a == b)  # Result: True  (Value is same)
print(a is b)  # Result: False (Type is different)

# 2. Comparing a and c
print(a == c)  # Result: True  (Value is same)
print(a is c)  # Result: True  (Same memory spot)

# 3. Comparing b and c
print(b == c)  # Result: True  (Value is same)
print(b is c)  # Result: False (Type is different)
````

#### Challenge 1

Write a program which repeatedly reads numbers until the user enters "done". Once "done" is entered, print out the total, count, and average of the numbers. If the user enters anything other than a number, detect their mistake using try and except and print an error message and skip to the next number.

    Enter a number: 4
    Enter a number: 5
    Enter a number: bad data
    Invalid input
    Enter a number: 7
    Enter a number: done
    16 3 5.333333333333333
    
[PEW] E5.1

````python
# Initialize variables to track sum and count
total = 0.0
count = 0

while True:
    user_input = input("Enter a number: ")

    # 1. Clean the input and check for the exit word immediately
    # .strip() removes spaces, .lower() makes it case-insensitive
    if user_input.strip().lower() == 'done':
        break

    # 2. Try to process the input as a number
    try:
        number = float(user_input)

        # If successful, update the stats
        total = total + number
        count = count + 1

    except ValueError:
        # 3. If the user typed "bad data", catch the error and skip
        print("Invalid input")
        continue

# 4. Final output: Check if we actually have numbers to avoid dividing by zero
if count > 0:
    average = total / count
    print(total, count, average)
else:
    print("No numbers were entered.")
````

#### Challenge 2  

Write another program that prompts for a list of numbers as above and at the end prints out both the maximum and minimum of the numbers instead of the average.

[PEW] E5.2

````python
largest = None
smallest = None

while True:
    user_input = input("Enter a number: ")

    # Check for exit condition
    if user_input.lower() == 'done':
        break

    try:
        # Convert input to float
        num = float(user_input)

        # Update largest
        if largest is None or num > largest:
            largest = num

        # Update smallest
        if smallest is None or num < smallest:
            smallest = num

    except ValueError:
        print("Invalid input")
        continue

# Print the results
print(f"Maximum: {largest}")
print(f"Minimum: {smallest}")
````
